## **Ejercicio 1: Sorteo de equipos de laboratorio**
---
**Contexto:** En el laboratorio de la facultad hay 12 estudiantes y se necesita armar 4 equipos de 3 personas cada uno, de forma aleatoria, para la práctica de la próxima semana. El profesor quiere además poder verificar rápidamente que ningún estudiante quedó repetido en dos equipos (por error del sorteo).

**Consigna:** Dada la lista `estudiantes = ["Ana","Luis","Marta","Carlos","Sofía","Pedro","Laura","Diego","Valentina","Andrés","Camila","Felipe"]`:

1. Usa el módulo `random` (con `random.shuffle` o `random.sample`) para mezclar la lista de estudiantes.
2. Escribe una función `armar_equipos(lista, tamaño)` que reciba la lista mezclada y el tamaño de cada equipo, y devuelva un **diccionario** donde cada clave sea `"Equipo 1"`, `"Equipo 2"`, etc., y cada valor sea una **tupla** con los nombres de ese equipo.
3. Usando un ciclo `for`, imprime cada equipo con su nombre.
4. Convierte la lista original de 12 nombres en un array de NumPy y usa `np.unique` para verificar que no haya ningún nombre repetido (esto simula una validación rápida del sorteo).

**Debes usar en tu solución:** el módulo `random`, una función propia, un diccionario, una tupla y `np.unique`.

**Para explicarle al resto del curso:** ¿por qué usaron una tupla y no una lista para guardar cada equipo? ¿qué les mostró `np.unique` y cómo lo interpretan?

```python
import random
import sys
import subprocess

try:
    import numpy as np
except ModuleNotFoundError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "numpy"])
    import numpy as np

estudiantes = ["Ana","Luis","Marta","Carlos","Sofía","Pedro","Laura","Diego","Valentina","Andrés","Camila","Felipe"]

# 1. Mezclar (copia, para no perder la lista original)
mezclada = estudiantes.copy()
random.shuffle(mezclada)

# 2. Función que arma los equipos
def armar_equipos(lista, tamaño):
    equipos = {}
    for i in range(0, len(lista), tamaño):
        numero = i // tamaño + 1
        equipos[f"Equipo {numero}"] = tuple(lista[i:i + tamaño])
    return equipos

equipos = armar_equipos(mezclada, 3)

# 3. Imprimir cada equipo
for nombre, integrantes in equipos.items():
    print(nombre, "->", integrantes)

# 4. Verificación con np.unique
arr = np.array(estudiantes)
unicos = np.unique(arr)
print("\nTotal de nombres:", len(arr))
print("Nombres únicos:   ", len(unicos))
print("¿Sin repetidos?   ", len(arr) == len(unicos))
```

## **Ejercicio 2: Control de calidad en la línea de producción**
---
**Contexto:** Una fábrica de envases revisa el peso (en gramos) de cada lote producido en un turno. Un lote es válido si su peso está entre 495 y 505 gramos. Si algún lote pesa menos de 480 o más de 520 gramos, la máquina debe detenerse de inmediato por seguridad.

**Consigna:** Dado `pesos = np.array([498, 507, 500, 492, 501, 503, 510, 499, 485, 502])`:

1. Primero, usa **indexación booleana** para obtener directamente el array `pesos_aprobados = pesos[(pesos >= 495) & (pesos <= 505)]`, con los pesos que ya están dentro del rango.
2. Escribe una función `evaluar_lote(peso)` que devuelva `"Aprobado"` si el peso está entre 495 y 505, o `"Rechazado"` en caso contrario (usa condicionales).
3. Ahora usa `np.where` para clasificar **todos** los pesos a la vez (sin ciclo) y guarda el resultado en un array `estado`. Compara: ¿cuántos elementos tiene `pesos_aprobados` y cuántos `"Aprobado"` hay en `estado`? Deberían coincidir.
4. Recorre `estado` con un ciclo `for` y cuenta cuántos lotes fueron aprobados y cuántos rechazados; guarda el resultado en un diccionario `{"Aprobado": n, "Rechazado": m}`.
5. Usa `np.any` para verificar si hubo al menos un lote fuera del rango de seguridad (menor a 480 o mayor a 520).

**Debes usar en tu solución:** indexación booleana (`pesos[condición]`), una función propia, `np.where`, `np.any`, un diccionario y condicionales.

**Para explicarle al resto del curso:** ¿por qué el filtro booleano y `np.where` les dieron resultados consistentes entre sí? ¿qué les dijo `np.any` sobre la máquina?

```python
import numpy as np

pesos = np.array([498, 507, 500, 492, 501, 503, 510, 499, 485, 502])

# 1. Indexación booleana
pesos_aprobados = pesos[(pesos >= 495) & (pesos <= 505)]
print("Pesos aprobados:", pesos_aprobados)

# 2. Función con condicionales
def evaluar_lote(peso):
    if 495 <= peso <= 505:
        return "Aprobado"
    else:
        return "Rechazado"

# 3. np.where sobre todo el array (sin ciclo)
estado = np.where((pesos >= 495) & (pesos <= 505), "Aprobado", "Rechazado")
print("Estado:", estado)
print("Elementos en pesos_aprobados:", len(pesos_aprobados))
print("'Aprobado' en estado:        ", np.sum(estado == "Aprobado"))
print("¿Coinciden?", len(pesos_aprobados) == np.sum(estado == "Aprobado"))

# 4. Conteo con ciclo for y diccionario
conteo = {"Aprobado": 0, "Rechazado": 0}
for e in estado:
    conteo[str(e)] += 1
print("Conteo:", conteo)

# 5. np.any: ¿algún lote fuera del rango de seguridad?
fuera_seguridad = np.any((pesos < 480) | (pesos > 520))
print("¿Hay que detener la máquina?", fuera_seguridad)
```

## **Ejercicio 3: Reporte de calificaciones por grupo**
---
**Contexto:** Un profesor tiene las notas de 4 estudiantes en 3 cortes del semestre y quiere un reporte que calcule el promedio de cada uno y los clasifique en categorías.

**Consigna:** Dados `notas = np.array([[3.5,4.0,3.8],[2.0,2.5,2.8],[4.5,4.8,4.9],[3.0,2.8,3.2]])` y `nombres = ["Ana","Luis","Marta","Carlos"]`:

1. Calcula el promedio de cada estudiante (por fila) usando `axis=1`.
2. Usando el array `notas` completo (sin `axis`), calcula con `.max()` y `.min()` cuál fue la nota más alta y la más baja de **todo** el curso, sin importar de qué estudiante o de qué corte es.
3. Escribe una función `clasificar(promedio)` que devuelva `"Aprobado"` (promedio ≥ 3.0), `"En riesgo"` (entre 2.5 y 3.0) o `"Reprobado"` (menor a 2.5), usando `if` / `elif` / `else`.
4. Usa un ciclo `for` junto con `zip(nombres, promedios)` para recorrer nombre y promedio al mismo tiempo, y construye un diccionario donde las claves sean las 3 categorías y los valores sean **listas de tuplas** `(nombre, promedio)` de cada estudiante en esa categoría.
5. Imprime el reporte final, categoría por categoría, junto con la nota máxima y mínima del curso.

**Debes usar en tu solución:** array 2D con `axis`, `.max()`/`.min()` sobre todo el array, una función, condicionales, `zip`, tuplas y un diccionario de listas.

**Para explicarle al resto del curso:** ¿cómo se aseguraron de que cada promedio quedara asociado al nombre correcto? ¿qué diferencia hay entre el promedio por fila (`axis=1`) y el máximo de todo el array sin `axis`?

```python
import numpy as np

notas = np.array([[3.5,4.0,3.8],[2.0,2.5,2.8],[4.5,4.8,4.9],[3.0,2.8,3.2]])
nombres = ["Ana","Luis","Marta","Carlos"]

# 1. Promedio por estudiante (por fila)
promedios = notas.mean(axis=1)

# 2. Máximo y mínimo de todo el curso (sin axis)
nota_max = notas.max()
nota_min = notas.min()

# 3. Clasificación
def clasificar(promedio):
    promedio = round(promedio, 2)   # evita errores de decimales (3.0000000000000004)
    if promedio >= 3.0:
        return "Aprobado"
    elif promedio >= 2.5:
        return "En riesgo"
    else:
        return "Reprobado"

# 4. Diccionario de listas de tuplas
reporte = {"Aprobado": [], "En riesgo": [], "Reprobado": []}
for nombre, prom in zip(nombres, promedios):
    reporte[clasificar(prom)].append((nombre, float(round(prom, 2))))

# 5. Reporte final
for categoria, lista in reporte.items():
    print(f"{categoria}: {lista}")
print("\nNota más alta del curso:", nota_max)
print("Nota más baja del curso:", nota_min)
```

## **Ejercicio 4: Generador y validador de contraseñas**
---
**Contexto:** Un sistema debe generar 10 contraseñas temporales para usuarios nuevos y reportar cuántas cumplen con un mínimo de seguridad: al menos 8 caracteres y al menos un dígito.

**Consigna:**

1. Escribe una función `generar_password(longitud)` que use `random.choice` para construir, carácter por carácter, una contraseña de `longitud` caracteres combinando letras y números.
2. Usando un ciclo (o una comprensión de lista), genera una lista de 10 contraseñas, llamando a la función anterior con una longitud aleatoria entre 6 y 10 para cada una (`random.randint`).
3. Escribe una función `es_segura(password)` que devuelva `True` o `False` según si tiene longitud ≥ 8 y contiene al menos un dígito (puedes recorrer los caracteres de la contraseña con un ciclo).
4. Convierte en un array de NumPy la lista de resultados `True`/`False` de las 10 contraseñas, y usa `np.unique` con `return_counts=True` para contar cuántas son seguras y cuántas no.
5. Imprime cuántas de las 10 contraseñas generadas cumplen el mínimo de seguridad.

**Debes usar en tu solución:** el módulo `random`, dos funciones propias, condicionales, una lista y `np.unique`.

**Para explicarle al resto del curso:** ¿qué revisa exactamente la función `es_segura` en cada contraseña? ¿qué les mostró `np.unique` sobre el conjunto generado?

```python
import random
import string
import numpy as np

# 1. Generar una contraseña carácter por carácter
def generar_password(longitud):
    caracteres = string.ascii_letters + string.digits
    password = ""
    for _ in range(longitud):
        password += random.choice(caracteres)
    return password

# 2. Lista de 10 contraseñas con longitud aleatoria entre 6 y 10
passwords = [generar_password(random.randint(6, 10)) for _ in range(10)]

# 3. Validador
def es_segura(password):
    tiene_digito = False
    for c in password:
        if c.isdigit():
            tiene_digito = True
    if len(password) >= 8 and tiene_digito:
        return True
    else:
        return False

# 4. Resultados -> array de NumPy -> np.unique con conteos
resultados = np.array([es_segura(p) for p in passwords])
valores, conteos = np.unique(resultados, return_counts=True)
print("Contraseñas:", passwords)
print("Resultados: ", resultados)
print("np.unique ->", dict(zip(valores.tolist(), conteos.tolist())))

# 5. Cuántas cumplen el mínimo
seguras = dict(zip(valores.tolist(), conteos.tolist())).get(True, 0)
print(f"\n{seguras} de {len(passwords)} contraseñas cumplen el mínimo de seguridad.")
```

## **Ejercicio 5: Simulación de dados (Monte Carlo)**
---
**Contexto:** Un profesor de probabilidad quiere mostrar que, al lanzar un dado muchas veces, la frecuencia observada de cada cara se acerca a la probabilidad teórica (1/6 ≈ 16.7%).

**Consigna:**

1. Usa `np.random.seed(0)` y `np.random.randint(1, 7, size=600)` para simular 600 lanzamientos de un dado.
2. Escribe una función `contar_frecuencias(lanzamientos)` que recorra el array con un ciclo `for` y construya un **diccionario** `{cara: conteo}` para las caras 1 a 6 (hazlo "a mano", sin `np.unique`, para practicar el conteo con ciclos).
3. Verifica tu conteo manual comparándolo con el resultado de `np.unique(lanzamientos, return_counts=True)`: deben coincidir.
4. Calcula el porcentaje observado de cada cara (conteo / 600 * 100, redondeado con `round`) y compáralo con el valor teórico de 100/6.
5. Usando un condicional, determina cuál cara se alejó **más** del valor teórico esperado.

**Debes usar en tu solución:** `np.random`, una función con un ciclo, un diccionario, `np.unique` y condicionales.

**Para explicarle al resto del curso:** ¿coincidió su conteo manual con el de `np.unique`? ¿qué cara se alejó más de lo esperado, y a qué creen que se debe esa diferencia?

```python
import numpy as np

# 1. Simulación
np.random.seed(0)
lanzamientos = np.random.randint(1, 7, size=600)

# 2. Conteo "a mano" con ciclo for
def contar_frecuencias(lanzamientos):
    conteo = {cara: 0 for cara in range(1, 7)}
    for valor in lanzamientos:
        conteo[int(valor)] += 1
    return conteo

manual = contar_frecuencias(lanzamientos)
print("Conteo manual:", manual)

# 3. Verificación con np.unique
caras, conteos = np.unique(lanzamientos, return_counts=True)
con_unique = dict(zip(caras.tolist(), conteos.tolist()))
print("Con np.unique:", con_unique)
print("¿Coinciden?", manual == con_unique)

# 4. Porcentajes observados vs. teórico
teorico = 100 / 6
porcentajes = {cara: round(n / 600 * 100, 2) for cara, n in manual.items()}
print(f"\nValor teórico: {round(teorico, 2)}%")
for cara, p in porcentajes.items():
    print(f"Cara {cara}: {p}%  (diferencia: {round(p - teorico, 2)})")

# 5. Cara que más se alejó del valor teórico
cara_max = None
desviacion_max = -1
for cara, p in porcentajes.items():
    desviacion = abs(p - teorico)
    if desviacion > desviacion_max:
        desviacion_max = desviacion
        cara_max = cara
print(f"\nLa cara que más se alejó fue la {cara_max} (desviación de {round(desviacion_max, 2)} puntos).")
```

## **Ejercicio 6: Itinerario de mantenimiento de equipos**
---
**Contexto:** Una planta tiene 5 máquinas, cada una con la fecha de su última revisión. Una máquina debe revisarse cada 30 días; si han pasado más días que eso, está "Vencida".

**Consigna:**

1. Define una lista de tuplas `equipos = [("Máquina A", date(2026,9,10)), ("Máquina B", date(2026,9,25)), ("Máquina C", date(2026,8,15)), ("Máquina D", date(2026,9,30)), ("Máquina E", date(2026,9,1))]` (usa `from datetime import date`).
2. Escribe una función `dias_transcurridos(fecha)` que calcule cuántos días han pasado desde `fecha` hasta hoy (`date.today() - fecha`, y usa `.days` para obtener el número).
3. Recorre `equipos` con un ciclo `for`; para cada máquina calcula los días transcurridos y, con un condicional, clasifícala como `"A tiempo"` o `"Vencida"` (más de 30 días).
4. Guarda los resultados en un diccionario `{"A tiempo": [...], "Vencida": [...]}` con los nombres de las máquinas en cada lista.
5. Reúne todos los "días transcurridos" en un array de NumPy y calcula el promedio, el máximo (la máquina más atrasada) y el **mínimo** (la máquina revisada más recientemente) de días sin revisión entre todas las máquinas.

**Debes usar en tu solución:** el módulo `datetime`, tuplas, una función, condicionales, un diccionario y NumPy (`mean`, `max`, `min`).

**Para explicarle al resto del curso:** ¿cómo calcularon los días transcurridos a partir de una fecha? ¿qué máquina corresponde al mínimo y cuál al máximo, y qué deberían hacer con cada una?

```python
from datetime import date
import numpy as np

# 1. Lista de tuplas
equipos = [("Máquina A", date(2026,9,10)), ("Máquina B", date(2026,9,25)),
           ("Máquina C", date(2026,8,15)), ("Máquina D", date(2026,9,30)),
           ("Máquina E", date(2026,9,1))]

# 2. Días transcurridos desde la última revisión
def dias_transcurridos(fecha):
    return (date.today() - fecha).days

# 3 y 4. Clasificación y diccionario
resultado = {"A tiempo": [], "Vencida": []}
dias = []
for nombre, fecha in equipos:
    d = dias_transcurridos(fecha)
    dias.append(d)
    if d > 30:
        resultado["Vencida"].append(nombre)
    else:
        resultado["A tiempo"].append(nombre)
print("Hoy:", date.today())
print(resultado)

# 5. NumPy sobre los días transcurridos
dias = np.array(dias)
nombres = [e[0] for e in equipos]
print("\nDías sin revisión:", dias)
print("Promedio:", dias.mean())
print("Máximo:", dias.max(), "->", nombres[int(np.argmax(dias))], "(la más atrasada)")
print("Mínimo:", dias.min(), "->", nombres[int(np.argmin(dias))], "(la revisada más recientemente)")
```

## **Ejercicio 7: Inventario y reorden automático de bodega**
---
**Contexto:** Una bodega registró las ventas semanales de 3 productos durante 4 semanas. Se debe decidir cuáles productos reabastecer según su promedio de ventas y el stock actual.

**Consigna:** Dados `ventas = np.array([[20,25,22,30],[5,6,4,5],[40,38,42,45]])` (filas = productos, columnas = semanas), `productos = ["Cuadernos","Lápices","Cafés"]` y `stock_actual = {"Cuadernos":15,"Lápices":50,"Cafés":10}`:

1. Calcula el promedio semanal de ventas de cada producto usando `axis=1`.
2. Imprime `ventas.T` (la matriz transpuesta) y observa que ahora cada **fila** es una semana y cada **columna** es un producto. Usando `ventas.T`, calcula con `axis=1` el total vendido por la bodega en cada semana, y confirma que te da el mismo resultado que `ventas.sum(axis=0)` sobre la matriz original.
3. Llega un cuarto producto a la bodega: `nuevas_ventas = np.array([12, 15, 10, 14])` ("Marcadores"). Usa `np.vstack` para agregar esta fila a la matriz `ventas`, y agrega `"Marcadores"` a la lista `productos` y su stock al diccionario `stock_actual`.
4. Escribe una función `necesita_reorden(promedio, stock, umbral=20)` que devuelva `True` si el promedio de ventas supera `umbral` **y** el stock actual es menor al doble del promedio (usa condicionales).
5. Recorre con un ciclo `for` la lista de productos (ya con los 4); para cada uno, consulta su stock en el diccionario, llama a la función anterior, y si el resultado es `True`, agrega una **tupla** `(producto, promedio, stock)` a una lista `a_reabastecer`.
6. Imprime la lista final de productos que necesitan reorden.

**Debes usar en tu solución:** array 2D con `axis`, `.T`, `np.vstack`, una función, condicionales, un diccionario, un ciclo `for` y una lista de tuplas.

**Para explicarle al resto del curso:** ¿qué cambia (y qué no) al transponer la matriz de ventas? ¿por qué un producto puede tener ventas altas y, aun así, no necesitar reorden?

```python
import numpy as np

ventas = np.array([[20,25,22,30],[5,6,4,5],[40,38,42,45]])
productos = ["Cuadernos","Lápices","Cafés"]
stock_actual = {"Cuadernos":15,"Lápices":50,"Cafés":10}

# 1. Promedio semanal por producto
print("Promedios:", ventas.mean(axis=1))

# 2. Transpuesta: total por semana
print("\nventas.T:\n", ventas.T)
total_semana_T = ventas.T.sum(axis=1)
total_semana = ventas.sum(axis=0)
print("Total por semana (con .T y axis=1):", total_semana_T)
print("Total por semana (axis=0 original):", total_semana)
print("¿Iguales?", np.array_equal(total_semana_T, total_semana))

# 3. Nuevo producto
nuevas_ventas = np.array([12, 15, 10, 14])
ventas = np.vstack([ventas, nuevas_ventas])
productos.append("Marcadores")
stock_actual["Marcadores"] = 8      # stock de ejemplo (la consigna no lo da)
promedios = ventas.mean(axis=1)     # recalculado con los 4 productos

# 4. Regla de reorden
def necesita_reorden(promedio, stock, umbral=20):
    if promedio > umbral and stock < 2 * promedio:
        return True
    return False

# 5. Recorrido
a_reabastecer = []
for i, producto in enumerate(productos):
    stock = stock_actual[producto]
    if necesita_reorden(promedios[i], stock):
        a_reabastecer.append((producto, float(round(promedios[i], 2)), stock))

# 6. Resultado
print("\nProductos a reabastecer (producto, promedio, stock):")
print(a_reabastecer)
```

## **Ejercicio 8: Comparador de rutas de reparto**
---
**Contexto:** Un repartidor tiene varias rutas posibles para visitar las mismas paradas en distinto orden. Quiere saber, antes de salir, cuál ruta recorre menos distancia en total.

**Consigna:** Define al menos 3 rutas como listas de tuplas de coordenadas, por ejemplo `ruta_1 = [(0,0),(3,4),(6,8),(9,2)]`, `ruta_2 = [(0,0),(6,8),(3,4),(9,2)]`, `ruta_3 = [(0,0),(9,2),(3,4),(6,8)]`:

1. Escribe una función `distancia(p1, p2)` que calcule la distancia entre dos puntos usando `math.sqrt((x2-x1)**2 + (y2-y1)**2)` (usa `import math`).
2. Escribe una función `distancia_total(ruta)` que recorra la lista de tuplas con un ciclo `for`, sumando la distancia entre cada par de paradas consecutivas (usando la función anterior), y devuelva el total.
3. Calcula la distancia total de cada una de las 3 rutas y guárdalas en un array de NumPy.
4. Usa `np.sort` para ver las 3 distancias ordenadas de menor a mayor, y usa `.min()` y `.max()` sobre ese mismo array para saber cuál es la ruta más corta y la más larga en distancia.
5. Usa `np.argmin` sobre el array original (sin ordenar) para identificar el **índice** de la ruta más corta, y `np.round` para mostrar las 3 distancias con 2 decimales.
6. Imprime cuál ruta conviene tomar y su distancia total.

**Debes usar en tu solución:** el módulo `math`, tuplas, dos funciones propias, un ciclo `for` y NumPy (`sort`, `min`, `max`, `argmin`, `round`).

**Para explicarle al resto del curso:** ¿qué diferencia hay entre `np.sort` (que ordena los valores) y `np.argmin` (que da el índice del menor)? ¿por qué después de ordenar ya no sabían cuál ruta era cuál?

```python
import math
import numpy as np

ruta_1 = [(0,0),(3,4),(6,8),(9,2)]
ruta_2 = [(0,0),(6,8),(3,4),(9,2)]
ruta_3 = [(0,0),(9,2),(3,4),(6,8)]

# 1. Distancia entre dos puntos
def distancia(p1, p2):
    x1, y1 = p1
    x2, y2 = p2
    return math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

# 2. Distancia total de una ruta
def distancia_total(ruta):
    total = 0
    for i in range(len(ruta) - 1):
        total += distancia(ruta[i], ruta[i + 1])
    return total

# 3. Array con las distancias de las 3 rutas
distancias = np.array([distancia_total(r) for r in (ruta_1, ruta_2, ruta_3)])

# 4. Ordenar y min/max
ordenado = np.sort(distancias)
print("Ordenadas:", np.round(ordenado, 2))
print("Más corta:", round(ordenado.min(), 2), "| Más larga:", round(ordenado.max(), 2))

# 5. Índice de la ruta más corta y redondeo
mejor = int(np.argmin(distancias))
print("Distancias (originales):", np.round(distancias, 2))

# 6. Conclusión
print(f"\nConviene la Ruta {mejor + 1}, con una distancia total de {np.round(distancias[mejor], 2)}.")
```

## **Ejercicio 9: Encuesta de satisfacción de clientes**
---
**Contexto:** Un restaurante recogió calificaciones (escala de 0 a 10) de 12 clientes y quiere clasificarlos al estilo NPS: Promotor (9-10), Pasivo (7-8) o Detractor (0-6).

**Consigna:** Dados `calificaciones = np.array([9,7,10,5,8,6,9,10,3,7,8,9])` y `clientes = ["Cliente " + str(i) for i in range(1, 13)]`:

1. Primero, usa **indexación booleana** para obtener directamente el array `promotores = calificaciones[calificaciones >= 9]`, con las calificaciones de 9 o 10.
2. Escribe una función `clasificar_cliente(nota)` que devuelva `"Promotor"`, `"Pasivo"` o `"Detractor"` según los rangos anteriores, usando condicionales.
3. Usando un ciclo `for` con `zip(clientes, calificaciones)`, construye un diccionario `{"Promotor": [...], "Pasivo": [...], "Detractor": [...]}` donde cada lista contenga tuplas `(cliente, nota)`. Verifica que la cantidad de elementos en `reporte["Promotor"]` coincida con el tamaño de `promotores` del paso 1.
4. Usa `np.unique(calificaciones, return_counts=True)` para identificar cuál calificación fue la más frecuente.
5. Calcula qué porcentaje de los 12 clientes son Promotores.

**Debes usar en tu solución:** indexación booleana (`calificaciones[condición]`), un diccionario de listas, tuplas, una función, condicionales y `np.unique`.

**Para explicarle al resto del curso:** ¿por qué el tamaño de `promotores` (indexación booleana) coincide con la cantidad de elementos en `reporte["Promotor"]`? ¿qué calificación fue la más frecuente según `np.unique`, y qué significa para el restaurante?

```python
import numpy as np

calificaciones = np.array([9,7,10,5,8,6,9,10,3,7,8,9])
clientes = ["Cliente " + str(i) for i in range(1, 13)]

# 1. Indexación booleana
promotores = calificaciones[calificaciones >= 9]

# 2. Clasificación NPS
def clasificar_cliente(nota):
    if nota >= 9:
        return "Promotor"
    elif nota >= 7:
        return "Pasivo"
    else:
        return "Detractor"

# 3. Diccionario de listas de tuplas
reporte = {"Promotor": [], "Pasivo": [], "Detractor": []}
for cliente, nota in zip(clientes, calificaciones):
    reporte[clasificar_cliente(nota)].append((cliente, int(nota)))
for categoria, lista in reporte.items():
    print(f"{categoria} ({len(lista)}): {lista}")
print("\n¿len(reporte['Promotor']) == promotores.size?", len(reporte["Promotor"]) == promotores.size)

# 4. Calificación más frecuente
valores, conteos = np.unique(calificaciones, return_counts=True)
mas_frecuente = valores[np.argmax(conteos)]
print("Más frecuente:", mas_frecuente, "con", conteos.max(), "clientes")

# 5. Porcentaje de promotores
porcentaje = len(reporte["Promotor"]) / len(calificaciones) * 100
print(f"Promotores: {round(porcentaje, 2)}%")
```

## **Ejercicio 10: Clima semanal y alerta por ola de calor**
---
**Contexto:** Una estación meteorológica registra la temperatura 3 veces al día (mañana, tarde, noche) durante una semana. Se considera que hubo "ola de calor" en un día si **las 3** lecturas de ese día superaron los 32°C.

**Consigna:**

1. Usa `np.random.seed(17)` y `np.random.randint(18, 38, size=21)` para simular, en un array **plano** de 21 lecturas, las temperaturas de toda la semana (3 lecturas por día × 7 días).
2. Usa `.reshape(7, 3)` para reorganizar ese array plano en una matriz de 7 días (filas) × 3 franjas (columnas).
4. Calcula el promedio de temperatura de cada día usando `axis=1`.
5. Escribe una función `hubo_ola_de_calor(lecturas_dia)` que reciba las 3 lecturas de un día (una fila del array) y devuelva `True` si **todas** superan 32°C, usando `np.all`.
6. Recorre los 7 días con un ciclo `for`, aplicando la función a cada fila, y construye un diccionario `{"Día 1": True/False, "Día 2": True/False, ...}`.
7. Usa `np.any` sobre los resultados para indicar si hubo ola de calor en **algún** momento de la semana.

**Debes usar en tu solución:** `np.random`, `.reshape`, un array 2D con `axis`, una función que use `np.all`, un ciclo `for`, un diccionario y `np.any`.

**Para explicarle al resto del curso:** ¿qué diferencia hay entre usar `np.any` y `np.all` en este ejercicio? ¿en qué día(s), según su diccionario, hubo ola de calor?

```python
import numpy as np

# 1. Simulación (array plano de 21 lecturas)
np.random.seed(17)
temperaturas = np.random.randint(18, 38, size=21)

# 2. 7 días x 3 franjas
semana = temperaturas.reshape(7, 3)
print(semana)

# 3. Promedio por día
print("\nPromedio por día:", np.round(semana.mean(axis=1), 2))

# 4. Función con np.all
def hubo_ola_de_calor(lecturas_dia):
    return bool(np.all(lecturas_dia > 32))

# 5. Diccionario por día
olas = {}
for i, dia in enumerate(semana, start=1):
    olas[f"Día {i}"] = hubo_ola_de_calor(dia)
print("\n", olas)

# 6. ¿Hubo ola de calor en algún momento?
print("¿Hubo ola de calor en la semana?", np.any(list(olas.values())))
```
